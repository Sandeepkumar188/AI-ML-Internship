```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
df = pd.read_csv("vacant_posts_2026-02-07.csv")

df.head()
```

```python
print("Number of rows and columns:", df.shape)
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
# Remove columns that are completely empty
df = df.dropna(axis=1, how="all")

# Remove duplicate rows
df = df.drop_duplicates()

print("Dataset cleaned successfully!")
print("New shape:", df.shape)
```

```python
print("Column names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)
```

```python
# Statistical summary of numeric columns
df.describe()
```

```python
df["Designation"].value_counts()
```

```python
df["Office Name"].value_counts()
```

```python
df["Division"].value_counts()
```

```python
designation_counts = df["Designation"].value_counts()

designation_counts.plot(kind="bar", figsize=(8, 5))
plt.title("Number of Posts by Designation")
plt.xlabel("Designation")
plt.ylabel("Number of Posts")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
division_counts = df["Division"].value_counts()

division_counts.plot(kind="bar", figsize=(9, 5))
plt.title("Number of Posts by Division")
plt.xlabel("Division")
plt.ylabel("Number of Posts")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
designation_counts = df["Designation"].value_counts()

designation_counts.plot(
    kind="pie",
    autopct="%1.1f%%",
    figsize=(7, 7)
)

plt.title("Percentage Distribution of Designations")
plt.ylabel("")
plt.tight_layout()
plt.show()
```

```python
df.to_csv("cleaned_vacant_posts.csv", index=False)

print("Cleaned CSV saved successfully!")
```
